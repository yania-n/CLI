```python
import matplotlib.pyplot as plt
```

```python
from PIL import Image
from transformers import pipeline
pipe = pipeline("image-classification",
"tonyassi/celebrity-classifier")
# https://huggingface.co/tonyassi/celebrity-classifier
```

Image classification of an input image:

```python
def predict(img):
  pred = pipe(img)
  print("Name:", pred[0]['label'])
  plt.imshow(img)
  plt.axis("off")
```

Image classification from an image dataset:

```python
from datasets import load_dataset, Image
dataset = load_dataset("ares1123/celebrity_dataset", split="train").cast_column("image", Image())
```
